TYPE CASTING

```python
#integer typecasting : int() function
v1=int(5.6) # float to int
v2=int("300") #string to int
#v3=int("hansika") #char string cant be converted,string should be convertable
v3=int(True) #Bool to int
v4=int("101",2) # binary to int("binary number",degree(number of digits for binary))

print(v1,type(v1))
print(v2,type(v2))
print(v3,type(v3))
print(v4,type(v4))
```

```python
#float typecasting : float() function
v1=float(5) # int to float
v2=float("300") #string to float
#v3=int("hansika") #char string can't be converted;  string should be convertible
v3=float(True) #Bool to int
#v4=float("101",2) binary cannot be converted into float

print(v1,type(v1))
print(v2,type(v2))
print(v3,type(v3))
print(v4,type(v4))
```

```python
#str typecasting : str() function
v1=str(5) # int to string
v2=str(300.5) #float to string
v3=str(True) #Bool to string
#v4=float("101",2) binary cannot be converted into float

print(v1,type(v1))
print(v2,type(v2))
print(v3,type(v3))
```

```python
#boolean typecasting : bool() function
v1=bool(5) # int to string
v2=bool(0) #float to string
v3=bool(True) #Bool to string; any other value other than 0 can be read as true
v5=bool("hansika")
#v4=float("101",2) binary cannot be converted into float

print(v1,type(v1))
print(v2,type(v2))
print(v3,type(v3))
print(v5,type(v5))
```
