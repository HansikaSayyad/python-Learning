```python
#receiving mutiple inputs-SPLIT() FUNCTION
x1, x2, x3=input("Enter three names").split()#input fucntion.split
#need to give the space(space makes the diff of inputs)
print(x1,x2,x3)
```

```python
#receiving mutiple inputs-SPLIT() FUNCTION
x1, x2, x3=input("Enter three names").split(',')#input fucntion.split
#need to give , as we nemtioned in split function(space makes the diff of inputs)
print(x1,x2,x3)
```

```python
x1, x2, x3=map(int,input("Enter three names").split(','))#map of int converting all inputs to some specific dataype
print(x1,x2,x3)
```

```python
x=list(map(int,input("Enter any number of number").split(',')))#enter any number of variables as list
print(x)
```

#Output in python

```python
#1.concatenation using commas
#2. String concatenation using +
#3.str.format() method
#4.f-strings methos(python 3.6+ version)
#5.escape sequences
#6.End and Sep parameters in print()
```

```python
#concatenation using commas
hero_name='prabhas'
age=40
fee=400
print(hero_name,age,fee)
```

```python
#2. String concatenation using +
hero_name='prabhas'
heroine_name='Anushka'
fee=100
print(hero_name+heroine_name)
print(hero_name+" " +heroine_name)
print(hero_name+" " +heroine_name+" " + str(fee))#all variables datatypes should be in same datatype
```

```python
#3.str.format() method
hero='NTR'
age=40
print("Name: {}     Age: {}".format(hero,age))
print("Name: {1}    Age: {0}".format(hero,age))
print("Name: {1}    Age: {0}".format(age,hero))
```

```python
#4.f-strings methos(python 3.6+ version)
hero='NTR'
age=40
print(f"Name:{hero}")
print(f"Name:{hero},Age:{age}")
```

```python
#5.escape sequences
hero_name='prabhas'
heroine_name='Anushka'
fee=100
print(f"heroname:{hero_name},\t heroinename:{heroine_name},\t FEE:{fee}")
print(f"heroname:{hero_name},\n heroinename:{heroine_name},\n FEE:{fee}")
print(f"heroname:\"{hero_name}\",\t heroinename:{heroine_name},\t FEE:{fee}")
```

```python
#6.End and Sep parameters in print()
hero_name='prabhas'
heroine_name='Anushka'
fee=100
print (hero_name,end="@")
print(heroine_name)
print(hero_name,heroine_name,fee,sep="#")#by default space
```
